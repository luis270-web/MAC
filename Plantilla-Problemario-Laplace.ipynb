```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import sympy
from scipy import integrate

# Configuración de notación matemática
sympy.init_printing(use_latex='mathjax')

# Variables y función
t = sympy.symbols("t", positive=True)
s = sympy.symbols("s", real=True)
y = sympy.Function("y")

# Defino la EDO
edo = y(t).diff(t, 2) + y(t) - sympy.sqrt(2)*sympy.sin(sympy.sqrt(2)*t)

# Transformada de Laplace de la EDO
L_edo = sympy.laplace_transform(edo, t, s, noconds=True)

# Expresión en el dominio s (considerando y(0) = 0 y y'(0) = 0 según la simplificación de la hoja)
f_s = 2 / ((s**2 + 1)*(s**2 + 2))

# Transformada Inversa de Laplace para obtener la solución y(t)
solucion = sympy.inverse_laplace_transform(f_s, s, t)
sympy.Eq(y(t), solucion)
```
